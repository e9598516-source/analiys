```python
import numpy as np 
import pandas  as pd 
import matplotlib.pyplot as plt 
```

```python
df=pd.DataFrame({
    "name":["Ali","Reza","Mohamad","Nima","Ehsan"],
    "age":[18,20,18,17,23],
    "mathScore":[12.5,15,14.25,18,10],
    "englishScore":[18,19,16,12.5,18.75],
    "pythonScore":[15,18,15.5,18.75,19.75],
})
df.head()
```

```python
english_mean=np.mean(df["englishScore"])
mathScore_mean=np.mean(df["mathScore"])
pythonScore_mean=np.mean(df["pythonScore"])
print(f"Lessons averag English={english_mean}  MathScore={mathScore_mean} PythonScore={pythonScore_mean}")
```

```python
english_min=np.min(df["englishScore"])
mathScore_min=np.min(df["mathScore"])
pythonScore_min=np.min(df["pythonScore"])
print(f"Lessons Min English={english_min}  MathScore={mathScore_min} PythonScore={pythonScore_min}")

```

```python
average=(df["englishScore"]+df["mathScore"]+df["pythonScore"])/3
```

```python

df.loc[average>15,"name"]

```

```python

plt.bar(df["name"],average )
plt.show()
```
